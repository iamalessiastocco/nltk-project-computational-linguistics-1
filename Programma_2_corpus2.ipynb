```python
import nltk
import math # per il calcolo di MI

from nltk import punkt # importa il modulo punkt da NLTK, che è utilizzato per la tokenizzazione delle frasi
from collections import Counter # da una libreria python si importa la classe Counter per contare le frequenze
from nltk.corpus import wordnet # importa il corpus wordnet da NLTK, che è una grande base di dati lessicale di inglese
from nltk import ngrams, FreqDist # importa ngrams per estrarre bigrammi e trigrammi e FreqDist per creare automaticamente la distribuzione di frequenza di una lista
from nltk.corpus import stopwords, words
from nltk import Maxent_NE_Chunker

nltk.download('averaged_perceptron_tagger_eng') # scarica il tagger POS per l'inglese
nltk.download('punkt_tab') # scarica un modello di tokenizzazione più efficace
nltk.download('wordnet')
nltk.download('stopwords')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
c1 = 'corpora_2.txt' # assegnamento alla variabile c1 del file da analizzare
with open (c1,"r") as infile: # apertura del file in modalità lettura
  contenuti_c1 = infile.read() # il contenuto dei file viene memorizzato nella variabile

# sentence splitting
from nltk.tokenize import sent_tokenize # dalla libreria NLTK si è presa la funzione sent_tokenize
sentences_c1 = sent_tokenize(contenuti_c1) # la funzione prende in input i file e li suddivide in frasi
# le frasi ottenute vengono poi salvate nelle variabili sentence_c1 e sentence_c2
print("sentences_corpus_1: ", sentences_c1)

# tokenizzazione
from nltk.tokenize import word_tokenize # dalla libreria NLTK si prende la funzione word_tokenize
tokens_c1 = word_tokenize(contenuti_c1) # la funzione prende in input il file e lo suddivide in token
# i risultati vengono salvati nella variabile tokens_c1
print("tokens_corpus_1: ", tokens_c1)

# pos tagging
pos_c1 = nltk.tag.pos_tag(tokens_c1) # la funzione prende in input la lista di token del file e restituisce una lista di tuple (word, POS tag) salavete nella variabile pos_c1
# questa funzione assegna dei tag ad ogni token in base alla loro categoria grammaticale
print("pos_corpus_1: ", pos_c1)

# estrazione dei top-50 sostantivi, avverbi e aggettivi più frequenti (con relativa frequenza, ordinata per frequenza decrescente)

# si crea la seguente funzione che ha come parametri una lista di tuple, ciascuna contenente un token e il suo relativo tag POS,
# una lista di tag POS che si vogliono filtrare e il numero di token più comuni da restituire, con un valore predefinito di 50
def estrazione_top_50_pos(pos_tags, categorie, top_n=50):
    # si filtrano i token in base alle categorie specificate nel parametro categorie
    # il filtraggio avviene tramite comprensione di lista che itera su pos_tags e mantiene solo i token il cui tag POS è presente in categorie
    token_filtrati = [token for token, pos in pos_tags if pos in categorie]
    # si calcola la frequenza di ciascun token
    frequenze = FreqDist(token_filtrati)
    # si ordina i token per frequenza decrescente e seleziona i 50 più frequenti
    top_token = frequenze.most_common(top_n)
    return top_token

# si richiama la funzione appena creata per individuare i 50 aggettivi, verbi e sostantivi più frequenti
top_sostantivi = estrazione_top_50_pos(pos_c1, ['NN', 'NNS', 'NNP', 'NNPS'], top_n=50)
top_avverbi = estrazione_top_50_pos(pos_c1, ['RB', 'RBR', 'RBS'], top_n=50)
top_aggettivi = estrazione_top_50_pos(pos_c1, ['JJ', 'JJR', 'JJS'], top_n=50)

# stampa dei risultati
print("Top 50 Sostantivi: ", top_sostantivi)
print("Top 50 Avverbi: ", top_avverbi)
print("Top 50 Aggettivi: ", top_aggettivi)

# estrazione dei top-20 n-grammi più frequenti (con relativa frequenza, e ordinati per frequenza decrescente)

# si individuano i bigrammi e trigrammi a partire dai token del corpus
bigrammi = nltk.ngrams(tokens_c1, 2)
trigrammi = nltk.ngrams(tokens_c1, 3)

# creazione della seguente funzione che ha come parametri gli n-grammi di cui si vogliono estrarre i top 20 più frequenti
def estrazione_top_ngrammi(ngrammi, top_n=20):
    frequenze = FreqDist(ngrammi) # si calcola la frequenza di ciascun n-gram e li si ordinano in modo decrescente
    top_ngrammi = frequenze.most_common(top_n)
    return top_ngrammi

# chiamata alla funzione per applicarla a unigrammi, bigrammi e trigrammi
top_unigrammi = estrazione_top_ngrammi(tokens_c1, top_n=20)
top_bigrammi = estrazione_top_ngrammi(bigrammi, top_n=20)
top_trigrammi = estrazione_top_ngrammi(trigrammi, top_n=20)

# stampa dei risultati
print("Top 20 Unigrammi: ", top_unigrammi)
print("Top 20 Bigrammi: ", top_bigrammi)
print("Top 20 Trigrammi: ", top_trigrammi)

# estrazione dei top 20 n-grammi di PoS più frequenti (con relativa frequenza, e ordinati per frequenza decrescente)

# estraggo a partire dalla lista di tuple pos, word solo i pos e utilizzo la funzione di prima
# per estrarre i top 20 ordinati in frequenza decrescente
solo_pos = [pos for word, pos in pos_c1]
bigrammi_pos = nltk.ngrams(solo_pos, 2)
trigrammi_pos = nltk.ngrams(solo_pos, 3)
quattrogrammi_pos = nltk.ngrams(solo_pos, 4)
cinquegrammi_pos = nltk.ngrams(solo_pos, 5)

top_unigrammi_pos = estrazione_top_ngrammi(solo_pos, top_n=20)
top_bigrammi_pos = estrazione_top_ngrammi(bigrammi_pos, top_n=20)
top_trigrammii_pos = estrazione_top_ngrammi(trigrammi_pos, top_n=20)
top_quattrogrammi_pos = estrazione_top_ngrammi(quattrogrammi_pos, top_n=20)
top_cinquegrammi_pos = estrazione_top_ngrammi(cinquegrammi_pos, top_n=20)

print("Top 20 Unigrammi PoS: ", top_unigrammi_pos)
print("Top 20 Bigrammi PoS: ", top_bigrammi_pos)
print("Top 20 Trigrammi PoS: ", top_trigrammii_pos)
print("Top 20 Quattrogrammi PoS: ", top_quattrogrammi_pos)
print("Top 20 Cinquegrammi PoS: ", top_cinquegrammi_pos)

# estrazione dei top-10 bigrammi composti da verbo e sostantivo
bigrammi_verbo_sostantivo = [
    (word1, word2)
    for (word1, pos1), (word2, pos2) in nltk.bigrams(pos_c1)
    if pos1.startswith('V') and pos2.startswith('NN')
]

# stampa dei 10 bigrammi in ordine di frequenza decrescente con relativa funzione utilizzando
# la funzione creata in precedenza di cui si è modificato il valore di default da 20 a 10
print("Top 10 bigrammi verbo sostantivo: ", estrazione_top_ngrammi(bigrammi_verbo_sostantivo, top_n=10))

# stampa dei 10 bigrammi in ordine di probabilità condizionata massima con relativo valore di probabilità

# calcolo delle frequenze dei bigrammi verbo - sostantivo e delle parole w1 in questi bigrammi
frequenze_bigrammi = nltk.FreqDist(bigrammi_verbo_sostantivo)
frequenze_parole_1 = nltk.FreqDist([word1 for word1, word2 in bigrammi_verbo_sostantivo])
frequenze_parole_2 = nltk.FreqDist([word2 for word1, word2 in bigrammi_verbo_sostantivo])

# si calcola la probabilità condizionata per ogni bigramma verbo - sostantivo
probabilita_condizionata = [] # si inizializza una lista vuota dove poi verranno inseriti i bigrammi e
                              # la loro probabilità condizionata corrispondente
for bigramma, freq_bigramma in frequenze_bigrammi.items(): # si inizializza un ciclo for che itera su tutti gli elementi di frequenze_bigrammi
    word1, word2 = bigramma # si scompone la tupla bigramma in word1, word2 così da poter recuperare poi solo la frequenza di word1
    freq_word1 = frequenze_parole_1[word1]
    prob_condizionata = freq_bigramma / freq_word1 # calcolo della probabilità condizionata
    probabilita_condizionata.append((bigramma, prob_condizionata)) # aggiunta alla lista del bigramma con probabilità associata

# ordinamento di lista tramite sorted secondo due criteri specifici: la funzione lambda (che estrae il valore
# da utilizzare per l'ordinamento e x[1] si riferisce al secondo elemento della tupla) e reverse
probabilita_condizionata_ordinata = sorted(probabilita_condizionata, key=lambda x: x[1], reverse=True)
top_10_bigrammi_prob_condizionata = probabilita_condizionata_ordinata[:10]
print("Top 10 bigrammi prob condizionata: ", top_10_bigrammi_prob_condizionata)

# stampa dei 10 bigrammi in ordine di probabilità congiunta massima con relativo valore di probabilità
totale_bigrammi = sum(frequenze_bigrammi.values())
probabilita_congiunta = []
for bigramma, freq_bigramma in frequenze_bigrammi.items():
    prob_congiunta = freq_bigramma / totale_bigrammi
    probabilita_congiunta.append((bigramma, prob_congiunta))

probabilita_congiunta_ordinata = sorted(probabilita_congiunta, key=lambda x: x[1], reverse=True)
top_10_bigrammi_prob_congiunta = probabilita_congiunta_ordinata[:10]
print("Top 10 bigrammi prob congiunta: ", top_10_bigrammi_prob_congiunta)

# stampa dei 10 bigrammi in ordine di MI massima, con relativo valore di MI
MI_results = []
for bigramma, freq_bigramma in frequenze_bigrammi.items():
    word1, word2 = bigramma
    freq_word1 = frequenze_parole_1[word1]
    freq_word2 = frequenze_parole_2[word2]
    # controllo che i valori di freq_word1 e freq_word2 siano diversi da 0 così da evitare l'errore ZeroDivisionError
    if freq_word1 == 0 or freq_word2 == 0:
        MI = 0
    else:
        MI = math.log((freq_bigramma * len(tokens_c1)) / (freq_word1 * freq_word2), 2)
    MI_results.append((bigramma, MI))

MI_results_ordinati = sorted(MI_results, key=lambda x: x[1], reverse=True)
top_10_bigrammi_MI = MI_results_ordinati[:10]
print("Top 10 bigrammi MI: ", top_10_bigrammi_MI)

# stampa dei 10 bigrammi in ordine di MI massima, con relativo valore di MI
LMI_results = []
for bigramma, MI in MI_results:
    freq_bigramma = frequenze_bigrammi[bigramma]
    LMI = freq_bigramma * MI
    LMI_results.append((bigramma, LMI))

LMI_results_ordinati = sorted(LMI_results, key=lambda x: x[1], reverse=True)
top_10_bigrammi_LMI = LMI_results_ordinati[:10]
print("Top 10 bigrammi LMI: ", top_10_bigrammi_LMI)

# calcolo e stampa del numero di elementi comuni ai top-10 per MI e per LMI
count = 0
for bigramma_mi, _ in top_10_bigrammi_MI:
    for bigramma_lmi, _ in top_10_bigrammi_LMI:
        if bigramma_mi == bigramma_lmi:
            count += 1
            break
print("Numero di elementi comuni tra MI e LMI: ", count)

# estrazione delle frasi con una lunghezza compresa tra 10 e 20 token, in cui almeno la metà
# dei token occorre almeno 2 volte nel corpus
corpus_counter = Counter(tokens_c1) # si contano le occorrenze di ciascun token nel corpus
frasi_filtrate = []
for frase in sentences_c1: # si tokenizzano le frasi del corpus, così poi da contare quanti token ci sono nelle frasi
    tokens_frase = word_tokenize(frase)
    if 10 <= len(tokens_frase) <= 20:
  # si contano quanti token distinti (set) delle frasi appaiono almeno due volte nel corpus
        count_repeated = sum(1 for token in set(tokens_frase) if corpus_counter[token] >= 2)
  # si calcola la metà del numero di token distinti nella frase
        half_tokens = len(set(tokens_frase)) // 2
  # si verifica che il numero di token distinti che appaiono due o più volte nel corpus
  # sia maggiore o uguale alla metà del numero di token distinti
        if count_repeated >= half_tokens:
          frasi_filtrate.append(frase)

# identificazione della frase con la media della distribuzione di frequenza dei token più alta
def media_frequenza(frase):
  tokens_frase = word_tokenize(frase)
  return sum(corpus_counter[token] for token in tokens_frase) / len(tokens_frase)

frase_frequenza_alta = max(frasi_filtrate, key=media_frequenza) # tramite key si specifica che la funzione verrà applicata ad ogni frase in frasi_filtrate
print("Frase con media della distribuzione di frequenza dei token più alta: ", frase_frequenza_alta)
frase_frequenza_bassa = min(frasi_filtrate, key=media_frequenza)
print("Frase con media della distribuzione di frequenza dei token più bassa: ", frase_frequenza_bassa)

# identificazione della frase con probabilità più alta secondo un modello di Markov di ordine 2 costruito
# a partire dal corpus di input
def markov2(frasi_filtrate, tokens_c1, corpus_size):
  count_trigrammi = FreqDist(nltk.trigrams(tokens_c1))
  count_bigrammi = FreqDist(nltk.bigrams(tokens_c1))
  count_unigrammi = FreqDist(tokens_c1)

  # si inizializza una probabilità uguale a 0 e una stringa vuota che
  # successivamente conterrà la frase con la probabilità più alta
  probabilita_frase_max = 0
  frase_probabilita_alta = ""

  for frase in frasi_filtrate: # si itera su ciascuna frase e la si tokenizza
    tokens_frase = word_tokenize(frase)
    probabilita_frase = 1.0 # inizializzazione di probabilità
    # si calcola la probabilità del primo token della frase
    first_token = tokens_frase[0]
    probabilita_frase *= count_unigrammi[first_token] / corpus_size
    # si calcola la probabilità del secondo token dato il primo
    second_token = tokens_frase[1]
    probabilita_frase *= count_bigrammi[(first_token, second_token)] / count_unigrammi[first_token]
    for i in range(2, len(tokens_frase)): # si itera sulla frase a partire dal terzo token (i=2)
      token_corrente = tokens_frase[i] # è il token di cui si vuole calcolare la probabilità
      due_token_precedenti = (tokens_frase[i-2], tokens_frase[i-1]) # si estraggono i due token precedenti al token corrente

      # si estrae la frequenza dei bigrammi e trigrammi
      freq_trigramma = count_trigrammi[(due_token_precedenti[0], due_token_precedenti[1], token_corrente)]
      freq_bigramma = count_bigrammi[due_token_precedenti]

      # calcolo della probabilita del token corrente che viene
      # moltiplicata con la probabilita della frase fino ad ora
      prob_token = freq_trigramma / freq_bigramma
      probabilita_frase *= prob_token

    if probabilita_frase > probabilita_frase_max:
      probabilita_frase_max = probabilita_frase
      frase_probabilita_alta = frase

  return frase_probabilita_alta

frase_probabilita_alta = markov2(frasi_filtrate, tokens_c1, len(tokens_c1))
print("Frase con probabilità più alta secondo un modello di Markov di ordine 2: ", frase_probabilita_alta)

# calcolo della percentuale di stopwords nel corpus rispetto al totale dei token
stopwords_result = []
for token in tokens_c1:
  if token in stopwords.words('english'):
    stopwords_result.append(token)
percentuale_stopwords = (len(stopwords_result) / len(tokens_c1)) * 100
print("Percentuale di stopwords nel corpus: ", percentuale_stopwords)

# calcolo del numero di pronomi personali sul totale di token
pronomi_personali = [pos for word, pos in pos_c1 if pos == 'PRP']
result = len(pronomi_personali) / len(tokens_c1)
print("Numero di pronomi personali sul totale dei token: ", result)

# calcolo del numero medio di pronomi personali per frase
def numero_medio_pronomi_per_frase(sentences_c1):
  pronomi_per_frase = []
  for frase in sentences_c1:
    tokens_frase = word_tokenize(frase)
    pronomi_frase_corrente = [pos for word, pos in nltk.pos_tag(tokens_frase) if pos == 'PRP']
    pronomi_per_frase.append(len(pronomi_frase_corrente))
  # la media si calcola sommando il totale dei pronomi e dividendola per il numero totale di frasi
  return sum(pronomi_per_frase) / len(sentences_c1)

result = numero_medio_pronomi_per_frase(sentences_c1)
print("Numero medio di pronomi personali per frase: ", result)

# estrazione delle entita nominate
albero_ne = nltk.ne_chunk(pos_c1)
NE = {}
for nodo in albero_ne:
  if hasattr(nodo, 'label'):
    entity_type = nodo.label()
    entity = "".join([token for token, pos in nodo.leaves()])
    if entity_type in NE.keys():
      NE[entity_type].append(entity)
    else:
      NE[entity_type] = [entity]

# identificazione per ciascuna classe di NE i 15 elementi più
# frequenti, ordinati per frequenza decrescente e con relativa frequenza
NE_frequenti_GPE = [FreqDist(NE['GPE']).most_common(15)]
NE_frequenti_PERSON = [FreqDist(NE['PERSON']).most_common(15)]
NE_frequenti_ORGANIZATION = [FreqDist(NE['ORGANIZATION']).most_common(15)]
print("NE 15 frequenti GPE: ", NE_frequenti_GPE)
print("NE 15 frequenti PERSON: ", NE_frequenti_PERSON)
print("NE 15 frequenti ORGANIZATION: ", NE_frequenti_ORGANIZATION)
```
